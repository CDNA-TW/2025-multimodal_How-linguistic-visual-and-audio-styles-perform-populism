```python
import pandas as pd

# 載入資料
df = pd.read_csv("all_results_summary_583.csv")
# df = pd.read_csv("all_results_summary_398.csv")
# 建立唯一參數鍵
df["param_key"] = df.apply(
    lambda row: f"{row['umap_n_neighbors']}-{row['umap_n_components']}-{row['umap_min_dist']}",
    axis=1
)

# 篩選 k 在 4~10 之間
df_filtered = df[df["k"].between(5, 8)]

# 每組參數中，取 silhouette 最大者
df_best_sil = df_filtered.loc[
    df_filtered.groupby("param_key")["silhouette_mean"].idxmax()
]

# 依 NMI 與 Silhouette 排序
df_ranked = df_best_sil.sort_values(
    by=["nmi_mean", "silhouette_mean"],
    ascending=[False, False]
).reset_index(drop=True)

# 顯示最後 10 筆
print(df_ranked.head(10))
```
